## **Bag of Words is a text representation technique where a document is converted into a vector based on word frequency, ignoring grammar and word order.**

```python
from sklearn.feature_extraction.text import CountVectorizer

sentences = [
    "I love NLP",
    "I love Python"
]

vectorizer = CountVectorizer()
bow = vectorizer.fit_transform(sentences)

print("Vocabulary:", vectorizer.get_feature_names_out())
print("BoW Matrix:\n", bow.toarray())
```

Disadvantages of Bag of Words (BoW)

- Ignores word order, so sentence meaning can be lost.
- No semantic understanding of words or their relationships.
- Cannot handle negation properly (e.g., not good vs good).
- Creates high-dimensional sparse vectors when vocabulary is large.
- Treats similar words as different (e.g., car and cars).
- Does not capture context, so same word in different meanings looks identical.
- Scales poorly with large text corpora.

For example :

```python
sentences = [
    "I love this movie",
    "I love this movie not"
]

vectorizer = CountVectorizer()
bow = vectorizer.fit_transform(sentences)

print(vectorizer.get_feature_names_out())
print(bow.toarray())
```

Here above, Both vectors are almost identical, but meanings are opposite.

```python
sentences = [
    "car",
    "cars"
]

vectorizer = CountVectorizer()
bow = vectorizer.fit_transform(sentences)

print(vectorizer.get_feature_names_out())
print(bow.toarray())
```

Both of the above words have same meaning but the vectors are different
