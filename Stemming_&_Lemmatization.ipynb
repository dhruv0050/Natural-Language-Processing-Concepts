# **What is stemming?**

Stemming is a text preprocessing technique that reduces words to their root form.
The root word may not always be a meaningful or correct dictionary word.

```python
import nltk
```

```python
words = ["running", "studies", "played", "playing", "better", "cats", "happiness"]
```

**Porter Stammer**

```python
from nltk.stem import PorterStemmer
porter = PorterStemmer()

for word in words:
    print(word + "-->" + porter.stem(word))
```

**Lancaster Stemmer**

```python
from nltk.stem import LancasterStemmer
lancaster = LancasterStemmer()

for word in words:
    print(word + "-->" + lancaster.stem(word))
```

**Regexp Stemmer**

```python
from nltk.stem import RegexpStemmer
regexp = RegexpStemmer('ing$|s$|e$|able$', min = 4)

for word in words:
    print(word + "-->" + regexp.stem(word))
```

**Snowball Stemmer**

```python
from nltk.stem import SnowballStemmer
snowball = SnowballStemmer('english', ignore_stopwords=False)

for word in words:
    print(word + "-->" + snowball.stem(word))
```

# **What is Lemmatization?**
Lemmatization is a text preprocessing technique that converts words into their base dictionary form.
It considers the word’s meaning and grammatical context to produce meaningful results.

```python
import nltk
nltk.download('wordnet')

# it takes an extra argument of pos that defines the nature of that word {'v' = Verb, 'a' = Adjective, 'n' = Noun, 'r' = Adverb}.
from nltk.stem import WordNetLemmatizer
lemma = WordNetLemmatizer()
```

```python
for word in words:
    print(word + "-->" + lemma.lemmatize(word, pos = 'v'))
```

```python
for word in words:
    print(word + "-->" + lemma.lemmatize(word, pos = 'n'))
```

```python
for word in words:
    print(word + "-->" + lemma.lemmatize(word, pos = 'a'))
```

```python
for word in words:
    print(word + "-->" + lemma.lemmatize(word, pos = 'r'))
```
