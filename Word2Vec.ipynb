**Word2Vec is a word embedding technique that converts words into dense numerical vectors such that words with similar meanings have similar vectors.**
**Instead of counting words (like BoW/TF-IDF), Word2Vec learns meaning from context.**

It learns by looking at neighboring words.
Two models:
- CBOW: predict the target word from context
- Skip-Gram: predict context words from a target word

```python
from gensim.models import Word2Vec

sentences = [
    ["i", "love", "nlp"],
    ["i", "love", "machine", "learning"],
    ["nlp", "is", "fun"],
    ["machine", "learning", "is", "powerful"]
]

# Training Word2Vec model
model = Word2Vec(
    sentences,
    vector_size=50,
    window=2,
    min_count=1,
    sg=1   # sg=1 → Skip-Gram, sg=0 → CBOW
)

# Get vector for a word
vector = model.wv["nlp"]
print("Vector for 'nlp':", vector)

```

Why Word2Vec is better than BoW / TF-IDF :
- Captures meaning
- Dense vectors (not sparse)
- Similar words have similar vectors
