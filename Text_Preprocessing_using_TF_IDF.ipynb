## **TF-IDF (Term Frequency - Inverse Document Frequency)**

TF-IDF is a text representation technique that assigns importance to words based on:
- how often a word appears in a document (TF)
- how rare the word is across all documents (IDF)

### **TF-IDF Formula**

---
### **Term Frequency (TF)**
Measures how often a word appears in a document.

TF(t, d) =( Number of rep of words in sentence ) / ( Total number of words in sentence )

---

### **Inverse Document Frequency (IDF)**

Measures how important a word is across all documents.


IDF(t) = loge[ (No. of sentences) /(No. of sentences containing the word) ]

---

### **TF-IDF**

Final score is the product of TF and IDF.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I love NLP",
    "I love Python",
    "I love love love"
]

vectorizer = TfidfVectorizer()
tfidf = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())
print(tfidf.toarray())
```

**Disadvantages of TF-IDF**
- Ignores word order, so sentence meaning can be lost.
- No semantic understanding of words (cannot capture synonyms).
- Cannot handle context or polysemy (same word, different meaning).
- Fails with negation, similar to Bag of Words.
- Produces sparse, high-dimensional vectors for large vocabularies.
- Rare but unimportant words may get high weight.
- Not suitable for deep contextual understanding required in modern NLP tasks.
